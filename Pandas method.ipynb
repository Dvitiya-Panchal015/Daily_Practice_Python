```python
#11. Create class BankAccount with private __balance. Add deposit and withdraw
# methods. Raise custom exception if balance becomes negative.


class NegativeBalanceError(Exception):
    def __init__(self, message="Balance cannot be negative!"):
        super().__init__(message)


class BankAccount:
    def __init__(self,balance=0):
        self.__balance = balance
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}.current balance:{self.__balance}")
        else:
            print(f"Deposit amount must be greater than 0.")
    def withdraw(self,amount):
        if amount <= 0:
            print("withdraw amount must be greater than 0")
        elif amount > self.__balance:
            raise NegativBalanceError("Insufficient funds:")
        else:
            self.__balance -= amount
            print(f"withdraw {amount}")
    def get_balance(self):
        return self.__balance
try:
    acc = BankAccount(20000)
    acc.deposit(500)
    acc.withdraw(800)
except NegativeBalanceError as e:
    print("Exception:",e)
```

```python
#14.Create class Rectangle with private length and width. Add method to calculate
# area and perimeter.

class Rectangle:
    def __init__(self,length,width):
        self.__length = length
        self.__width = width
    def area(self):
        return self.__length * self.__width
    def perimeter(self):
        return 2*(self.__length + self.__width)
    def get_length(self):
        return self.__length
    def get_width(self):
        return self.__width
r = Rectangle(23,5)

print(r.area())
print(r.perimeter())
print(r.get_length())
print(r.get_width())
```

# Inheritance + Abstraction + super() (15-19)

```python
#15. Create single level inheritance: Person → Student. Use super() in constructor.

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def show_info(self):
        print(f"name: {self.name}, age: {self.age}")
class Student(Person):
    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.marks = marks
    def show_info(self):
        print(f"marks: {self.marks}")

p = Person("Shweta",34)
p.show_info()

s = Student("Shweta",34,99)
s.show_info()
```

```python
#16. Create multi-level inheritance: Person → Employee → Manager. Override a
# method at each level.

class Person:
    def __init__(self,name):
        self.name = name
    def show(self):
        print(f"Person name :{self.name}")
class Employee(Person):
    def __init__(self,name,salary):
        self.salary = salary
        super().__init__(name)
    def show(self):
        print(f"Employee salary:{self.salary}")
class Manager(Employee):
    def __init__(self,name,salary,department):
        self.department = department
        super().__init__(name,salary)
    def show(self):
        print(f"Manager department : {self.department}")
        
e = Employee("Mahi",40000)
e.show() 

m = Manager("Mahi",30000,"IT")
m.show()
```

```python
#17. Create hierarchical inheritance: Parent Animal → Dog and Cat. Override sound()
# method.

class Animal:
    def sound(self):
        print("Animal sound:")
class Dog(Animal):
    def sound(self):
        print("Dog sound bhauuu...")
class cat(Animal):
    def sound(self):
        print("Cat sound meowww...")

d = Dog()
d.sound()

c = cat()
c.sound()
```

```python
#18. Create abstract class Shape with abstract method area(). Implement it in Circle
# and Square using super().

from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self,color):
        self.color = color
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius,color):
        super().__init__(color)
        self.radius = radius
    def area(self):
        return 3.14 * (self.radius ** 2)
class Square(Shape):
    def __init__(self,side,color):
        super().__init__(color)
        self.side = side
    def area(self):
        return self.side ** 2

c = Circle(6,"Red")
print(c.area())

s = Square(5,"Red")
print(s.area())
```

```python
#20. Create a class Calculator that can add 2 or 3 numbers (method overloading
# using *args). 

class Calculator:
    def add(self, *args):
        if len(args) == 2:
            return args[0] + args[1]
        elif len(args) == 3:
            return args[0] + args[2]
        else:
            raise ValueError("Provide either 2 or 3 numbers")
calc = Calculator()

print(calc.add(12,34))
print(calc.add(2,3,4))       
```

```python
#21. Create class Point and overload + and > operators.

class Point:
    def __init__(self,x,y):
       self.x = x
       self.y = y
    def __add__(self,other):
       return Point(self.x + other.x, self.y + other.y)
    def __gt__(self,other):
        self_dist = (self.x**2 + self.y**2) ** 0.5
        other_dist = (self.x**2 + self.y**2) ** 0.5
        return self_dist >other_dist
    def __str__(self):
        return f"Point({self.x}, {self.y})"
        
p1= Point(5,5)
p2= Point(7,7)
print("Addition:",p1+p2)
print("Is p1 > p2",p1>p2)
print("Is p1 >p2", p1>p2)
```

```python
#25. Create a menu-driven program for file operations: Write, Read, Append, and Exit.
# Handle FileNotFoundError.

while True:
    print("\nMenu:")
    print("1. Write to file")
    print("2. Read file")
    print("3. Append to file")
    print("4. Exit")

    choice = int(input("Enter your choice :"))

    if choice == 1:
        filename = input("Enter filename:")
        data = input("Enter text to write:")
        with open(filename,"w") as file:
            file.write(data + "\n")
        print("Data written sucessfully.")
    elif choice == 2:
        filename = input("Enter filename:")
        try:
            with open(filename,"r")as file:
                content = file.read()
                print("\n file content :\n",content)
        except FileNotFoundError:
            print("Error: File not found. Please write something first.")

    elif choice == 3:
        # Append content
        filename = input("Enter filename: ")
        try:
            data = input("Enter text to append: ")
            with open(filename, "a") as file:
                file.write(data + "\n")
            print("Data appended successfully.")
        except FileNotFoundError:
            print("Error: File not found. Please create/write file first.")

    elif choice == 4:
        print("Exiting program... Goodbye!")
        break

    else:
        print("Invalid choice! Please try again.")
```
