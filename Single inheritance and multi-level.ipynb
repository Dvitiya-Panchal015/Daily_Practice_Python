```python
#19. Class Name:
# Shape → Polygon → Rectangle
# Problem
# Create multi-level inheritance.
# Explanation
# Rectangle should calculate its area using length and width.

class Shape:
    def __init__(self, color):
        self.color = color
class polygon(Shape):
    def __init__(self,color,sides):
        self.sides = sides
        super().__init__(color)
class Rectangle(polygon):
    def __init__(self,color,length,width):
        self.length = length
        self.width  = width
        super().__init__(color,4) #when constructor call polygon then the reactangle have the 4 sides)
    def area(self):
        return self.length * self.width

rect = Rectangle("Blue", 10, 5)
print(f"Rectangle Color: {rect.color}")
print(f"Rectangle Sides: {rect.sides}")
print(f"Rectangle Area: {rect.area()}")      
```

```python
#29. Class Name:
# Shape → TwoDShape → Triangle
# Problem
# Create multi-level inheritance.
# Explanation
# Triangle should calculate its area using Heron's Formula.
# Store:
# • side1
# • side2
# • side3
# Implement the area calculation.

import math
class Shape:
    def __init__(self,name):
        self.name = name
    def display(self):
        print(f"shape name : {self.name}")
        
class TwoShape(Shape):
    def __init__(self,name):
        super().__init__(name)
    def display(self):
        super().display()
        print("this is a 2D Shape:")
class Triangle(TwoShape):
    def __init__(self,side1,side2,side3):
        super().__init__("Triangle")
        self.side1=side1
        self.side2 = side2
        self.side3 = side3
    def area(self):
        s = (self.side1+self.side2+self.side3)/2
        return math.sqrt(s * (s - self.side1)*(s-self.side2)*(s-self.side3))
    def display(self):
        super().display()
        print(f"sides : {self.side1},{self.side2},{self.side3}")
        print(f"area(Heron's Formula): {self.area():.2f}")

t = Triangle(2,3,4)
t.display()
```

```python
#30. Class Name:
# GrandParent → Parent → Child → GrandChild
# Problem
# Create four levels of inheritance.
# Explanation
# Each class should have its own method.
# The GrandChild object should be able to access methods from all parent classes.
# Demonstrate Python's Method Resolution Order (MRO) using:
# ClassName.mro()
# or
# help(ClassName)


class GrandParent:
    def grandparent_method(self):
        print("GrandParent Sharing wisdom.")

class Parent(GrandParent):
    def parent_method(self):
        print("Parent Providing guidance.")

class Child(Parent):
    def child_method(self):
        print("Child Learning and growing.")

class GrandChild(Child):
    def grandchild_method(self):
        print("GrandChild Exploring the world.")


gc = GrandChild()

gc.grandparent_method()
gc.parent_method()
gc.child_method()
gc.grandchild_method()

print("\nMethod Resolution Order (MRO):")
print(GrandChild.mro())
```
