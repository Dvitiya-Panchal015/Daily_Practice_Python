```python
#15. Take three sides of a triangle. Check if they form a valid triangle. If
#yes, also print if it is Equilateral, Isosceles, or Scalene.

A=10
B=10
C=10

if A+B>C:
    print("Valid Triangle:")
    if A==B==C:
        print("Equilateral:")
    elif A==B and B==C or C==A:
        print("Isosceles:")
    else:
        print("Scalene:")
else:
    print("Not a valid traingle:")
```

```python
#16. Take student marks in 5 subjects. Calculate percentage and give
#grade: 90+ → A+, 80-89 → A, 70-79 → B, 60-69 → C, below 60 → Fail.

marks1= int(input("Enter a marks1:"))
marks2= int(input("Enter a marks2:"))
marks3= int(input("Enter a marks3:"))
marks4= int(input("Enter a marks4:"))
marks5= int(input("Enter a marks5:"))

total = (marks1+marks2+marks3+marks4+marks5)
print(total)
percentage =total/500*100
print(percentage)

if percentage>90:
    print("A+")
elif percentage>80 and percentage<80:
    print("A")
elif percentage>70 and percentage<70:
    print("B")
elif percentage>60 and percentage<60:
    print("C")
else:
    print("Fail")


```
