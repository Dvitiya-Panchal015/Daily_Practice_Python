```python
#4. Shape → Circle, Rectangle
# Question
# Define class Shape with method area(). Create subclasses Circle and Rectangle.
# Concept
# • Parent provides common method.
# • Children may later override it.
# How to Solve
# 1. Create Shape.
# 2. Add area().
# 3. Create Circle.
# 4. Create Rectangle.
# 5. Call area().

class Shape:
    def area(self):
        print("calculating the area of shape:")
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * (self.radius **2)
class Reactangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width

c = Circle(6)
print(c.area())
r = Reactangle(10,10)
print(r.area())
```

```python
#24. Bank System
# Concept
# Three children inherit one parent.
# How to Solve
# 1. Parent → Account
# 2. Store balance.
# 3. Savings → interest.
# 4. Current → overdraft.
# 5. FixedDeposit → maturity.
# 6. Show each separately.


class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def show_balance(self):
        print(f"Current Balance: ₹{self.balance}")


class SavingsAccount(Account):
    def calculate_interest(self, rate=0.05):
        interest = self.balance * rate
        print(f"Savings Account Interest: ₹{interest}")
        return interest


class CurrentAccount(Account):
    def overdraft(self, amount):
        if amount <= self.balance + 5000:   
            print(f"Overdraft Approved: ₹{amount}")
        else:
            print("Overdraft Limit Exceeded!")


class FixedDeposit(Account):
    def maturity_amount(self, years, rate=0.06):
        maturity = self.balance * ((1 + rate) ** years)
        print(f"Fixed Deposit Maturity after {years} years: ₹{maturity:.2f}")
        return maturity


s = SavingsAccount(20000)
c = CurrentAccount(30000)
f = FixedDeposit(50000)

s.show_balance()
s.calculate_interest()

c.show_balance()
c.overdraft(32000)

f.show_balance()
f.maturity_amount(5)   
```

```python
#27. Employee Salary
# Concept
# Different salary calculations.
# How to Solve
# 1. Parent stores common data.
# 2. FullTime calculates monthly salary.
# 3. PartTime calculates hourly salary.
# 4. Override calculation method. 

class Employee:
    def __init__(self,name):
        self.name = name
    def calculate_salary(self):
        print("Salary is calculated:")
class Fulltime(Employee):
    def __init__(self,name,monthly_salary):
        self.monthly_salary = monthly_salary
        super().__init__(name)
    def calculate_salary(self):
        print("Calculate the monthly salary:")

class Parttime(Employee):
    def __init__(self,name,hours_worked, hourly_rate):
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate
        super().__init__(name)
    def calculate_salary(self):
        salary = self.hours_worked * self.hourly_rate
        print(f"part time employee : {self.name} earns {salary} for {self.hours_worked} hours.")
        return salary

p = Parttime("sam",55,677)
p.calculate_salary()
```
