```python
#Temperature Converter
# Convert temperature from Celsius to Fahrenheit and vice versa.
# Formula: F = (C * 9/5) + 32

# Convert Celsius to Fahrenheit
celsius = 37
fahrenheit = (celsius * 9/5) + 32
print(celsius, "°C =", fahrenheit, "°F")

# Convert Fahrenheit to Celsius
fahrenheit = 98.6
celsius = (fahrenheit - 32) * 5/9
print(fahrenheit, "°F =", celsius, "°C")
```

```python
# Area and Perimeter Calculator
# Take radius of circle from user and calculate:
# Area and Circumference of circle
# (Use pi = 3.14)

radius = float(input("Enter radius of circle:"))
pi = 3.14
area = pi*radius*radius
circumference = 2*pi*radius

print("Area of cicle:",area)
print("Circumference of circle:",circumference)
```

```python
# Student Grade Calculator
# Take marks of 5 subjects.
# Calculate percentage and print grade:
# >=90 → A+, >=80 → A, >=70 → B, >=60 → C, else → Fail

sub1=float(input("Enter a subject1:"))
sub2=float(input("Enter a subject2:"))
sub3=float(input("Enter a subject3:"))
sub4=float(input("Enter a subject4:"))
sub5=float(input("Enter a subject5:"))

total=sub1+sub2+sub3+sub4+sub5
percentage= total/5

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >=70:
    grade = "B"
elif percentage >=60:
    grade = "C"
else:
    grade = "Fail"
print("Grade:", grade)
```
