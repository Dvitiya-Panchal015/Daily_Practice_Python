```python
#23. (Hard) Take 4 student names and their marks as input from the user. Store them in a dictionary (name as key, marks as value). Then find and print the student with the highest marks.
student = {}
for i in range(4):
    name = input("Enter the name :")
    marks = input("Enter the Marks:")
    student[name] = marks

print(student)
highest_marks = max(student,key = student.get)
highest_marks_std = student[highest_marks]

print(f"Highest marks student name is {highest_marks}{highest_marks_std}")

```
