```python
#14. Create class BankAccount with private attribute __balance. Add deposit and withdraw
# methods with validation.

class BankAccount:
    def __init__(self,balance=0):
        self.__balance = balance
    def deposit(self,amt):
        if amt>0:
            self.__balance += amt
            print(f"Desposited {amt} current balance : {self.__balance}")
        else:
            print("deposit amount must be greater than zero")
    def withdraw(self,amt):
        if amt <= 0:
            print(f"withdraw amount must be greater than zero:")
        elif amt > self.__balance:
            print("Insfficient balance:")
        else:
            self.__balance -= amt
            print(f"withdraw {amt}. Remaining balance : {self.__balance}")
    def get_balance(self):
        return self.__balance
a = BankAccount(1000)
a.deposit(300)
a.withdraw(500)

        
```

```python
#18. Create class Rectangle with private length and width. Add getter, setter and method to
# calculate area.

class Rectangle:
    def __init__(self,length,width):
        self.__length = length
        self.__width = width
    def get_len(self):
        return self.__length
    def get_wid(self):
        return self.__width
    def set_len(self,length):
        self.__length = length
    def set_wid(self,width):
        self.__width = width
    def area(self):
        return self.__length * self.__width
        
r = Rectangle(4,5)
r.area()
```

```python
#20. Create multi-level inheritance: Person → Employee → Manager.

class Person:
    def __init__(self,name):
        self.name = name 
    def display(self):
        print("Person name :",{self.name})
class Employee(Person):
    def __init__(self,name,emp_id):
        self.emp_id = emp_id
    def display(self):
        print("employee id:",{self.emp_id})
class Manager(Employee):
    def __init__(self,name,emp_id,dept):
        self.dept = dept
    def display(self):
        print("Manager department :",{self.dept})
e = Employee("Vidha",101)
e.display()

m = Manager("Vidha",101,"IT")
m.display()

p = Person("Vidha")
p.display()
```

```python
#23. Create abstract class Shape with abstract method area(). Implement it in Circle and
# Rectangle using super().

from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self,color):
        self.color = color
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius,color):
        super().__init__(color)
        self.radius = radius
    def area(self):
        return 3.14 * (self.radius**2)
class Rectangle(Shape):
    def __init__(self,length,width,color):
        super().__init__(color)
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width 
c = Circle(6,"Red")
print(c.area())

r = Rectangle(5,7,"Red")
print(r.area())
    
```

```python
#27. Create class Point and overload + operator to add two points.

class Point:
    def __init__(self,x,y):
        self.x = x 
        self.y = y
    def __add__(self,other):
        return Point(self.x + other.x,self.y + other.y)
    def display(self):
        print(f"{self.x} , {self.y}")
p1 = Point(3,4)
p2 = Point(4,5)
p3 = p1+p2
p3.display()
```

```python
#28. Create class Book and overload > operator to compare number of pages of two books.

class Book:
    def __init__(self,title,pages):
        self.title = title
        self.pages = pages
    def __gt__(self,other):
        return self.pages > other.pages
    def __str__(self):
        return f"book : {self.title}, pages:{self.pages}"
b1 = Book("Python",600)
b2 = Book("java", 500)

print(b1)
print(b2)

if b1 > b2:
    print(f"{b1.title} has more pages than {b2.title}")
else:
    print(f"{b2.title} has more pages than {b1.title}")
    
```

```python
#29. Create two different classes Duck and Person both having method walk(). Demonstrate
# Duck Typing.

class Duck:
    def walk(self):
        print("Duck is walking:")
class Person:
    def walk(self):
        print("Person walks on the road:")
def make_it_walk(obj):
    obj.walk()
d = Duck()
p = Person()

make_it_walk(d)
make_it_walk(p)
```

```python
#30. Create class ComplexNumber and overload both + and - operators.

class ComplexNumber:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return ComplexNumber(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return ComplexNumber(self.x - other.x, self.y - other.y)

    def display(self):
        print(f"{self.x} + {self.y}i")
        
p1 = ComplexNumber(3,8)
p2 = ComplexNumber(4,5)

print("Addition:", p1+p2)
print("Subtraction:",p1-p2)

p3.display()
```
