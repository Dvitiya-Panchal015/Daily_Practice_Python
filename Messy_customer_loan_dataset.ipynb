# Data Type & Invalid Values

```python
# 11.The Age column contains numbers as well as values like "unknown" and
# "twenty five". Clean this column and convert it into a proper numeric column.


df["age"] = pd.to_numeric(df["age"],errors = "coerce")
df["age"]
```

```python
# 12.The Annual Income column contains values such as 650000 INR, ₹720000,
# 1.2L, and N/A. Clean the column and convert the valid values into numeric
# format.

df["annual_income"] = df["annual_income"].astype(str)
df["annual_income"]
```

```python
df["aanual_income"] = df["annual_income"].str.replace("INR"," ")
df["annual_income"] = df["annual_income"].str.replace("," , " ")
df["annual_income"] = df["annual_income"].str.replace("₹", " ")
df
```

```python
df["annual_income"] = df["annual_income"].str.strip()
df["annual_income"]
```

```python
df["annual_income"] = pd.to_numeric(df["annual_income"],errors = "coerce")
df["annual_income"]
```

```python
#13.Clean the Credit Score column and identify values that cannot be converted
# into numbers.

df["credit_score"] = df["credit_score"].astype(str)
df["credit_score"]
```

```python
df["credit_score"] = df["credit_score"].str.strip()
df["credit_score"]
```

```python
df["credit_score"] = df["credit_score"].str.replace("pts"," ")
df["credit_score"]
```

```python
df["credit_score"] = df["credit_score"].str.replace("score:"," ")
df
```

```python
df["credit_score"] = pd.to_numeric(df["credit_score"],errors = "coerce")
df
```

```python
# 14.The Loan Amount column contains currency symbols, text, and numeric
# values. Remove the unwanted characters and convert it into numeric format.

df["loan_amount_(inr)"] = df["loan_amount_(inr)"].str.replace("INR"," ")
df["loan_amount_(inr)"] = df["loan_amount_(inr)"].str.replace("," , " ")
df["loan_amount_(inr)"] = df["loan_amount_(inr)"].str.replace("₹", " ")
df["loan_amount_(inr)"] = df["loan_amount_(inr)"].str.replace("USD", " ")
df["loan_amount_(inr)"] = df["loan_amount_(inr)"].str.strip()

df
```

```python
df["loan_amount_(inr)"] = pd.to_numeric(df["loan_amount_(inr)"],errors = "coerce")
df
```

```python
#15.The Monthly Visits column contains values such as "five", "N/A", and "-".
# Convert valid values into numeric values and handle invalid values.

df["monthly_visits"] = df["monthly_visits"].astype(str).str.strip().str.lower()
df
```

```python
convert = {"one":1,"two":2,"three":3,"four": 4,"five":5}
df["monthly_visits"] = df["monthly_visits"].replace(convert)
df
```

```python
df["monthly_visits"] = pd.to_numeric(df["monthly_visits"],errors ="coerce")
df
```

```python
#16.Clean the Debt to Income column where some values are written like 35%
# and 0.4x. Convert the valid values into a consistent numeric format.

df["debt_to_income"] = df["debt_to_income"].astype(str).str.strip().str.lower()
df
```

```python
df["debt_to_income"] = df["debt_to_income"].str.replace("%"," ")
df
```

```python
df["debt_to_income"] =df["debt_to_income"].str.replace("x"," ")
df
```

# Categorical Cleaning

```python
# 18.Find all unique values in the Employment-Type column.

df.columns
```

```python
df["employment-type"].unique()
```

```python
# 19.Standardize the Employment-Type column so that values such as salaried,
# SALARIED, and Salaried become one consistent value.

df["employment-type"] =df["employment-type"].replace({"salaried":"SALARIED"})
df["employment-type"]
```

```python
# 20.Clean the Education Level column and make values such as graduate,
# GRADUATE, and Graduate consistent.

df["education__level"] = df["education__level"].replace({"graduate":"GRADUATE"})
df["education__level"]
```

```python
# 21.Clean the City column. Identify incorrect or inconsistent values such as
# hyderbad, Hyd, Bangalore, and Bengaluru.

df["city"].replace({"Hyderabad":"hyd",
                                 "Mumbai" :"mumbai",
                                 "Jaipur": "jaipur",
                                 "Kolkata" : "kolkata",
                                 "Delhi" :"delhi",
                                 "Pune" : "pune",
                                 "Chennai" :"chennai"},inplace = True)
df
```

```python
# 22.Find all unique values in the Customer Segment column and identify invalid
# categories.

df["customer_segment"] = df["customer_segment"].str.strip().str.lower()
df["customer_segment"]
```

```python
# 23.Clean the Loan Type column so that values such as personal, PERSONAL,
# and personal loan are standardized.

df["loan_type"].replace({
                         "Education" : "EDUCATION",
                         "Auto" : "AUTO",
                         "Home":"HOME",
                         "Home Loan":"HOME LOAN",
                        "Personal": "PERSONAL"},inplace = True)
df
```

```python
#24.Clean the Loan Approved column where values include Yes, yes, Y, 1, No,
#  NO, N, and 0. Convert them into only Yes/No.

df["loan_approved?"].replace({ "Yes":"yes",
                               "No" : "NO",
                               "Y" : 1,
                               "N" : 0},inplace = True)
df
```

# Outliers & Data Validation

```python
# 25.Check the Age column and identify unrealistic ages such as negative ages or
# ages above 100.

import numpy as np

df["age"] = [age if (0<= age <=120) else np.nan for age in df["age"]]
df["age"]

```

```python
# 26.Check the Credit Score column and identify values outside the valid credit-
# score range.
import numpy as np

df["credit_score"] = [score if (300 <= score <= 850) else np.nan for score in df["credit_score"]]
df["credit_score"]
```

```python
# 27.Detect outliers in Annual Income using the IQR method.



Q3 = df["annual_income"].quantile(0.25)
Q1 = df["annual_income"].quantile(0.25)

IQR = Q3 - Q1

lowerbond = Q1 - 1.5 * IQR
upperbond = Q3 + 1.5 * IQR

outliers = df[(df["annual_income"] < lowerbond) | (df["annual_income"] > upperbond)]
print(lowerbond)
print(upperbond)
print(outliers)
```

```python
# 28.Detect outliers in Loan Amount using the IQR method.

Q3 = df["loan_amount_(inr)"].quantile(0.25)
Q1 = df["loan_amount_(inr)"].quantile(0.25)

IQR = Q3 - Q1

lowerbond = Q1 - 1.5 * IQR
upperbond = Q3 + 1.5 * IQR

outliers = df[(df["loan_amount_(inr)"] < lowerbond) |(df["loan_amount_(inr)"]>upperbond)]
print(lowerbond)
print(upperbond)
print(outliers)
```
