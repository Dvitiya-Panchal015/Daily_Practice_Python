```python
# Problem 6
# Create a custom iterator that prints numbers from 1 to 10.

class numbers:
    def __init__(self):
        self.n= 1 
    def __iter__(self):
        return self
    def __next__(self):
        if self.n <= 10:
            x = self.n
            self.n += 1
            return x
        else:
            raise StopIteration
for num in numbers():
    print(num)
        
```

```python
# Create a custom iterator that prints numbers in reverse order from
# 10 to 1.

class numbers:
    def __init__(self):
        self.n = 10
    def __iter__(self):
        return self
    def __next__(self):
        if self.n >= 1:
            x = self.n 
            self.n -= 1
            return x
        else:
            raise StopIteration
for num in numbers():
    print(num)
```

```python
# Problem 9
# Create a custom iterator that prints the characters of the word
# "GENERATOR" one by one.

class word:
    def __init__(self,word):
        self.word = word
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch 
        else:
            raise StopIteration
for c in word("GENERATOR"):
    print(c)
```

```python
# Problem 16
# Given the following list:
# numbers = [10, 15, 20, 25, 30, 35]
# Create a generator that yields only the even numbers.

numbers = [10,15,20,25,30,35]
def Even_numbers():
    for i in numbers:
        if i % 2 == 0 :
            yield i
        
generator = Even_numbers()

for i in generator:
    print(i)
```

```python
# Problem 23
# Create a custom iterator that prints all alphabet letters from A to Z.

class word:
    def __init__(self):
        self.code = ord('A')
    def __iter__(self):
        return self
    def __next__(self):
        if self.code < ord('Z'):
            ch = chr(self.code)
            self.code += 1
            return ch 
        else:
            raise StopIteration
for c in word():
    print(c)
    
```

```python
# Problem 24
# Create a custom iterator that prints all vowels from the string below.

class vowel:
    def __init__(self,text):
        self.text = text
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        while self.index < len(self.text):
            ch = self.text[self.index]
            self.index += 1
            if ch.lower() in "aeiou":
                return ch 
        raise StopIteration
text = "Artificial Intelligence"
vowels = vowel(text)
for i in vowels:
    print(i)
```

```python
# Problem 25
# Create a custom iterator that prints the Fibonacci sequence up to 10  terms. 

class fibonacci:
    def __init__(self,terms):
        self.terms = terms
        self.count = 0
        self.a = 0
        self.b = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.count < self.terms:
            num = self.a
            self.a, self.b = self.b, self.a + self.b 
            self.count+= 1
            return num
        else:
            raise StopIteration
generator = fibonacci(10)

for i in generator:
    print(i)
```

# Level 6: Advanced Generator Problems 

```python
# Problem 26 
# Create a generator that yields the Fibonacci sequence up to 15  terms. 

def fibonacci(terms):
    a = 0
    b = 1

    for i in range(terms):
        yield a
        a, b = b, a + b


generator = fibonacci(15)

for i in generator:
    print(i)
```

```python
# Problem 27 
# Create a generator that yields all prime numbers between 1 and 100. 

def prime_num():
    num = 2
    while num <=100:
        is_prime = True 
        i = 2
        while i < num:
            if num % i == 0:
                is_prime = False
                break
            i += 1 
        if is_prime:
            yield num
        num += 1
generator = prime_num()

for i in generator:
    print(i)
```
