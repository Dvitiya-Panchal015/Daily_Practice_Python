# Basic Exploration & Missing Values

```python
# Basic Exploration & Missing Values
# 1. How many missing values are there in each column? Which column has the highest
# number of missing values?
# 2. What percentage of rows have at least one missing value?
# 3. Drop all rows that have more than 3 missing values. How many rows remain?



import pandas as pd 
import matplotlib as plt
import numpy as np
import seaborn as sns

df = pd.read_csv("C:\\Users\\Shree\\Downloads\\Customer_datasets.csv")
df
```

```python
print(df.isnull().sum(), "\nHighest:", df.isnull().sum().idxmax())
```

```python
print(df.isnull().any(axis=1).mean() * 100)
```

```python
df.dropna(thresh=df.shape[1] - 3)
```

# Duplicates

```python
# 4. How many exact duplicate rows are present in the dataset?

df.duplicated().sum()
```

```python
# 5. How many duplicate Customer_IDs are there? Keep only the first occurrence of each
# Customer_ID.


df["Customer_ID"] = df["Customer_ID"].duplicated().sum()
df["Customer_ID"]
```

# Age Column

```python
# 6. Convert the Age column to numeric. How many values could not be converted
# (became NaN)?

df["Age"] = pd.to_numeric(df["Age"],errors = "coerce")

df["Age"] = df["Age"].isna().sum()
df["Age"]
```

```python
# 7. After converting Age to numeric, identify and remove outliers using the IQR method.
# How many outliers were removed?


Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.25)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 *IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["Age"] < lower_bound ) | (df["Age"] > upper_bound)]

print(outliers)
print(lower_bound)
print(upper_bound)
```

```python
# 8. Replace remaining missing values in Age with the median age.

median_age = df["Age"].median()

df["Age"].fillna(median_age)
```

# Salary Column

```python
# 9. Clean the Salary column: remove currency symbols ($), commas, and convert
# everything possible to numeric. How many values remained non-numeric?

df["Salary"] = df["Salary"].replace({"$":"",
                                     ",":""})
df["Salary"]
```

```python
df["Salary"] = pd.to_numeric(df["Salary"],errors ="coerce")
df["Salary"]
```

```python
# 10. After cleaning, find the mean and median salary. Which one is more affected by
# outliers?

df["Salary"].median()
```

# Date Column

```python
# 11. Convert the Join_Date column to proper datetime format. How many dates failed to
# parse?


df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")
df["Join_Date"]
```

```python
# 12. Extract the year from the cleaned Join_Date. What is the most common joining year?

df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")
df["Join_Date"] = df["Join_Date"].dt.year.mode()[0]
df["Join_Date"]
```

# Categorical Columns (Inconsistencies)

```python
# 13. Standardize the Department column (fix casing, trim spaces, fix common typos like
# “Sale” → “Sales”, “Financ” → “Finance”, etc.). How many unique departments remain
# after cleaning?

df["Department"] = df["Department"].str.strip().str.title()
df["Department"] = df["Department"].replace({"Sale":"Sales",
                                             "Financ":"Finance",
                                             "Hr":"HR"})
df["Department"]
```

```python
# 14. Clean the City column (standardize names, fix “Los Angles” → “Los Angeles”, “NY” →
# “New York”, “Philly” → “Philadelphia”, etc.). List the final unique cities.

df["City"] = df["City"].replace({"Los Angles":"Los Angeles",
                                 "NY":"New York",
                                 "Philly":"Philandelphia"})
df["City"] = df["City"].nunique()
df["City"]
```

```python
# 15. Standardize the Gender column so it only contains “Male”, “Female”, and “Other”.
# How many records were changed?

df["Gender"] = df["Gender"].replace({
    "male": "Male", "m": "Male",
    "female": "Female", "f": "Female",
    "other": "Other", "o": "Other"
})
changed_count = (df["Gender"].isin(["Male","Female","Other"]) == False).sum()
changed_count
```

# Performance_Score & Is_Active

```python
# 16. Convert Performance_Score to numeric. Replace non-numeric values with NaN, then
# fill missing scores with the mean.

df["Performance_Score"] = pd.to_numeric(df["Performance_Score"],errors = "coerce")
df["Performance_Score"]
```

```python
df["Performance_Score"] = df["Performance_Score"].mean()
df["Performance_Score"]
```

```python
# 17. Clean the Is_Active column so it becomes a proper boolean (True/False). Map values
# like “Yes”, “Y”, “1”, “Active”, “true” → True and “No”, “N”, “0”, “Inactive”, “false” →
# False.

df["Is_Active"] = df["Is_Active"].replace({
    "yes": True, "y": True, "1": True, "active": True, "true": True,
    "no": False, "n": False, "0": False, "inactive": False, "false": False})
df["Is_Active"]
```

# Email & Name

```python
# 18. Flag invalid email addresses (those that do not contain “@” or have incomplete
# domains). How many invalid emails are there?

df["Email"] = df["Email"].astype(str).str.strip().str.contains("@") == False
df["Email"]
```

```python
# 19. Clean the Name column: replace empty strings, “N/A”, and “Unknown” with NaN.
# How many names are still missing after this step?

df["Name"] = df["Name"].replace(["", "N/A", "Unknown"], pd.NA)
df["Name"] = df["Name"].isna().sum()
df["Name"]
```

# Final Challenge

```python
# 20. Perform a complete data cleaning pipeline on the entire dataset and save the
# cleaned version as cleaned_dataset.csv. Report the final shape of the cleaned dataset
# and briefly describe the main cleaning steps you applied.

df = pd.read_csv("C:\\Users\\Shree\\Downloads\\Customer_datasets.csv")

df = df.drop_duplicates()

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

df = df.dropna(thresh=df.shape[1]-3)

df.to_csv("cleaned_dataset.csv", index=False)

print("Final shape:", df.shape)

df
```
