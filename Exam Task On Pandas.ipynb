# Part 5 — Categorical Cleaning

```python
#19. Clean City
# The City column contains values such as:
# Mumbai
# mumbai
# Mumbai
# mumbai
# Standardize the values so that logically identical cities are represented consistently.
# Then check the unique frequency distribution.
# Use: replace(), value_counts()


df["City"].value_counts()
```

```python
df["City"] = df["City"].replace({"mumbai":"Mumbai"})
df["City"]
```

```python
#20. Analyze Segment
# Check the frequency of:
# Retail
# SME
# Corporate
# Premium
# Also determine whether missing values exist.
# Then replace missing Segment values with the most appropriate category based on your
# analysis.
# Use: value_counts(), fillna()

df["Segment"] = df["Segment"].value_counts(dropna = False)
df["Segment"]
```

```python
df["Segment"] = df["Segment"].isna().sum()
df["Segment"]
```

```python
#21. Clean Active
# Active is supposed to be a binary column, but it contains:
# Yes
# No
# Y 
# N
# YES
# NO
# yes
# no
# 1
# 0
# Transform these into a consistent representation:
# Yes
# No
# Then verify the result with value_counts().
# Use: replace(), value_counts()

df["Active"] = df["Active"].replace({"Y":"Yes","N":"No","yes":1,"no":0})
df["Active"]
```

```python
df["Active"] = df["Active"].value_counts()
df["Active"]
```

```python
#22. Clean Risk_Level
# Check the distribution of:
# High
# Medium
# Low
# Verify whether missing values exist.
# Then display the final class distribution.

df["Risk_Level"] = df["Risk_Level"].isna().sum()
df["Risk_Level"]
```

```python
df["Risk_Level"] = df["Risk_Level"].value_counts()
df["Risk_Level"]
```
