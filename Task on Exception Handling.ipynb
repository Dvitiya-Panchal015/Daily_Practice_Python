```python
#20. Create a custom exception called NegativeNumberError and raise it when user enters a
# negative number.

class NegativeNumberError(Exception):
    def __init__(self,msg="number cannot be negative:"):
        super().__init__(msg)
try:
    num = int(input("Enter a no.:"))
    if num < 0:
        raise NegativeNumberError("you enter a negative no.")
    print("Valid number:")

except NegativeNumberError as e:
    print("Error:",e)
except ValueError:
    print("Error: please enter a valid integer:")
    
    
```

```python
#21.  Create a custom exception InsufficientBalanceError for a banking program.

class InsufficientBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self,balance):
        self.balance = balance
    def withdraw(self,amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Error: Insufficient balance:")
        self.balance -= amount
        print("Withdraw amount sucessfully:")
acc = BankAccount(50000)

try:
    acc.withdraw(3000)
except InsufficientBalanceError as e:
    print(e)
```

```python
#23. Create a function that checks password length and raises a custom exception if length is
# less than 8.

class PasswordTooShortError(Exception):
    def __init__(self,msg = "Password must be least 8 characters long:"):
        super().__init__(msg)

def check_pass(password):
    if len(password)<8:
        raise PasswordTooShortError("Password too short! Minimum 8 characters required.")
    else:
        print("Password accepted :",password)
try:
    p = input("Enter ur password:")
    check_pass(p)
except PasswordTooShortError as e:
    print ("Error:",e)
```

```python
#28. Create a banking system with deposit and withdraw. Raise and handle custom exception
# for insufficient balance.

class InsfficientBalanceError(Exception):
    def __init__(self,msg="Insufficent balance for withdrawal."):
        super().__init__(msg)
class BankAccount:
    def __init__(self,ac_no,balance = 0):
        self.ac_no = ac_no
        self.balance = balance
    def deposit(self,amount):
        if amount <= 0:
            print("Deposit amount must be positive:")
        else:
            self.balance += amount
            print(f"deposited {amount}. current balance : {self.balance}")
    def withdraw(self,amount):
        if amount <= 0:
            print("Withdrawal amount must be positive:")
        elif amount > self.balance:
            raise InsufficientBalanceError(f"Attempted to withdraw{amount}, but balane {self.balance}.")
        else:
            self.balance -= amount
            print(f"withdraw {amount}. Remaining Balance : {self.balance}")
    def get_balance(self):
        return self.balance
a = BankAccount(23434,5000)
a.deposit(3000)
a.withdraw(3000)
```

```python
#29. Take a list of numbers from user (as string). Convert them to integers and handle any
# conversion errors.



try:
    user_input = input("Enter numbers separated by spaces: ")

    str_numbers = user_input.split()

    int_numbers = []
    for s in str_numbers:
        try:
            int_numbers.append(int(s))
        except ValueError:
            print(f"Error: '{s}' is not a valid integer.")

    print("Final list of integers:", int_numbers)

except Exception as e:
    print("Unexpected error:", e)
```

```python
#30. Write a complete program that accepts user input for name, age and marks. Handle all
# possible exceptions and use finally to print “Program Ended”.


try:
    name = input("Enter a name:")
    age = int(input("Enter a age:"))
    if age < 0:
        raise ValueError("Age cannot be negative:")
    marks = float(input("Enter ur marks:"))
    if marks <0 or marks>100:
        raise ValueError("Marks must be between 0 and 100")
    print("name:",name)
    print("Age:",age)
    print("Marks:",marks)
except ValueError as e:
    print("Error:",e)
except Exception as e:
    print("Unexpected error:",e)
finally:
    print("\n Program Ended:")
```
