```python
#1.Create a Calculator class that can add 2 numbers or 3 numbers using the same method
# name.
# Concept
# Method Overloading (Python uses default arguments or *args to achieve overloading.)
# What to do
# • Create a class Calculator.
# • Create one method add().
# • The method should work for either 2 numbers or 3 numbers.
# How to solve
# 1. Create class Calculator.
# 2. Define method add().
# 3. Use default arguments or *args.
# 4. Check how many numbers are provided.
# 5. Return the sum.

class Calculator:
    def add(self,a,b,c=0):
        return a+b+c
c = Calculator()
print(c.add(10,10))

    
```

```python
#2. Create an Area class that calculates the area of a square and rectangle using the same
# method.
# Concept
# Method Overloading using default arguments.
# What to do
#• One parameter → Square.
# •  Two parameters → Rectangle.
# How to solve
# 1. Create class Area.
# 2. Define method calculate().
# 3. If only one value is provided, calculate the square's area.
# 4. If two values are provided, calculate the rectangle's area.
# 5. Return the result.

class Area:
    def calculate(self, length,width =0 ):
        
        if width == 0 :
            return  length * length
        else:
            return length * width 
a = Area()
print("square area:",a.calculate(7))
print("Rectangle area :", a.calculate(7,8))
```

```python
#3. Create a Student class with a details() method that accepts different numbers of
# arguments.
# Concept
# Method Overloading
# What to do
# The method should work with:
# • Name
# • Name + Age
# • Name + Age + City
# How to solve
# 1. Create class Student.
# 2. Create method details().
# 3. Use default arguments.
# 4. Print the information based on the provided values.

class Student:
    def details(self,name,Age = 0, city = 0):
        if Age == 0  and city == 0:
            print(f"name : {name}")
        elif city is None:
            print(f"name : {name}, Age : {Age}")
        else:
            print(f"name : {name}, Age : {Age}, city : {city}")
s = Student()
s.details("Raghav")
s.details("Raghav",23)
s.details("Raghav",34,"Hyd")
```

```python
#4. Create a MathOperations class that multiplies any number of values.
# Concept
# Method Overloading using *args
# What to do
# Multiply:
# • 2 numbers
# • 3 numbers
#• 4 numbers
# •  More numbers
# How to solve
# 1. Create class MathOperations.
# 2. Define method multiply(*args).
# 3. Initialize the answer as 1.
# 4. Loop through all numbers.
# 5. Multiply each number.
# 6. Return the final product.

class MathOperations:
    def multiply(self,*args):
        result = 1 
        for n in args:
            result += n 
        return result

m = MathOperations()
print(m.multiply(3,4,5) )  
print(m.multiply(4,5,6,6))
```

```python
#5. Create a Display class that prints different messages based on the argument type.
# Concept
# Method Overloading + Type Checking
# What to do
#Accept:
# •  Integer
# • String
# • List
# How to solve
# 1. Create class Display.
# 2. Create method show().
# 3. Use isinstance().
# 4. Print different messages depending on the object's type

class Display:
    def show(self,value):
        if isinstance(value,int):
            print (f"int value :{value}")
        elif isinstance(value,str):
            print(f"str value :{value}")
        elif isinstance(value,list):
            print(f"list value : {value}")
        else:
            print("type is not support")

d = Display()
d.show(20)
d.show("hello")
d.show([1,2,3])
```

```python
#6. Create a Volume class that calculates the volume of a cube and cuboid.
# Concept
# Method Overloading
# What to do
# • One parameter → Cube
# • Three parameters → Cuboid
# How to solve
# 1. Create class Volume.
# 2. Create method calculate().
# 3. If one value is given, calculate the cube's volume.
# 4. Otherwise, calculate the cuboid's volume.

class Volume:
    def Calculate(self,length, width = 0 , height = 0):
        if width ==0 and height ==0:
            return length **3
        elif width !=0 and height !=0:
            return length * width * height
        else:
            print("Either 1 value (cube) or value of cuboids)")

c = Volume()
print("cube value:",c.Calculate(5))
print("cuboid volume : ",c.Calculate(4,5,6))
```

```python
#7. Create a PrintData class that prints one or many values.
# Concept
# Method Overloading using *args
# What to do
# Print:
#• One value
# •  Multiple values
# How to solve
# 1. Create class PrintData.
# 2. Create method print_data(*args).
#  3. Loop through the arguments.
# 4. Print each value.

class PrintData:
    def print_data(self,*args):
        for i in args:
            print(i)
p = PrintData()
p.print_data(10)
p.print_data(10,20,30)
```

```python
# 8. Create an Animal class with sound(). Override it in Dog and Cat.
# Concept
# Method Overriding
# What to do
# • Animal → Generic sound
# • Dog → Bark
# • Cat → Meow
# How to solve
# 1. Create parent class Animal.
# 2. Define method sound().
# 3. Inherit Dog and Cat.
# 4. Override sound().
# 5. Create objects and call the method.

class Animal:
    def __init__(self,name):
        self.name = name
    def sound(self):
        print (f"{self.name} sound:")
   
class Cat(Animal):
    
    def sound(self):
        print(f"{self.name} sound : Meow Meow")
    
class Dog(Animal):
    
    def sound(self):
        print(f"{self.name} sound : bhauuu bhauuu")
    

def make_sound(animal,name):
    animal.sound(name)


c = Cat("Siri")
c.sound()

d = Dog("Ziggy")
d.sound()
```

```python
#9. Create Vehicle, Car, and Bike classes.
# Concept
# Method Overriding
# What to do
# Override the start() method.
# How to solve
# 1. Create Vehicle.
# 2. Add method start().
# 3. Create Car.
# 4. Create Bike.
# 5. Override start() in both child classes.

class Vehicle:
    def start(self):
        print("Vehicle is started:")
class Car(Vehicle):
    def start(self):
        print("Car is running:")
class Bike(Vehicle):
    def start(self):
        print("Bike is start:")
        
c = Car()
c.start()

b = Bike()
b.start()
```

```python
#10. Create an Employee class and override calculate_salary() in Manager.
# Concept
# Method Overriding
# What to do
# • Employee → Normal salary
# • Manager → Salary + Bonus
# How to solve
# 1. Create parent class.
# 2. Define calculate_salary().
# 3. Inherit Manager.
# 4. Override the method.
# 5. Print the manager's salary.


class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
        
    def calculate_salary(self):
        return self.salary
class Manager(Employee):
    def __init__(self,name,salary,bonus):
        self.bonus = bonus
        super().__init__(name,salary)
    def calculate_salary(self):
        return self.salary + self.bonus

m = Manager("Ram",25000,100)
print(f"{m.name} salary",m.calculate_salary())
```

```python
#11. Create Shape, Circle, and Rectangle classes.
# Concept
# Method Overriding
# What to do
# Override the area() method.
# How to solve
# 1. Create Shape.
# 2. Add area().
#3. Inherit Circle.
# 4 . Inherit Rectangle.
# 5. Override area() in each child class.

class Shape:
    def area(self):
        print("calculating the area of shapes:")
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * (self.radius ** 2)
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width 
s = Shape()
s.area()

c = Circle(3)
print(c.area())

r = Rectangle(5,4)
print(r.area())
```

```python
#13. Create Parent and Child classes using super().
# Concept
# Method Overriding + super()
# What to do
# Override display().
# How to solve
# 1. Create Parent.
# 2. Add display().
# 3. Create Child.
# 4. Override display().
# 5. Call super().display().
# 6. Print additional information.

class Parent:
    def display(self):
        print("this is parent class:")
class Child(Parent):
    def display(self):
        super().display()
        print("This is child class:")
p = Parent()
p.display()

c = Child()
c.display()
```

# Operator Overloading Questions (15-20)

```python
# 15. Create a Point class and overload the + operator.
# Concept
# Operator Overloading (__add__)
# What to do
# Add two points together.
# How to solve
# 1. Create class Point.
# 2. Store x and y coordinates.
# 3. Overload __add__().
# 4. Return a new point with added coordinates.

class Point:
    def __init__(self,x,y):
        self.x = x 
        self.y = y
    def __add__(self,other):
        return Point(self.x + other.x,self.y + other.y)
    def display(self):
        print(f"{self.x} , {self.y}")
p1 = Point(3,4)
p2 = Point(4,5)
p3 = p1+p2
p3.display()
```

```python
#16. Create a Book class and overload the + operator.
# Concept
# Operator Overloading
# What to do
# Add the number of pages of two books.
# How to solve
# 1. Create class Book.
# 2. Store pages.
# 3. Overload __add__().
# 4. Return the total pages.

class Book:
    def __init__(self,pages):
        self.pages = pages
    def __add__(self,other):
        return self.pages + other.pages
b1 = Book(200)
b2 = Book(900)

b3 = b1+b2
print("total:",b3)
```

```python
#17. Create a Student class and overload the > operator.
# Concept
# Comparison Operator Overloading (__gt__)
# What to do
# Compare marks of two students.
# How to solve
# 1. Create class Student.
# 2. Store marks.
# 3. Overload __gt__().
# 4. Compare the marks.
#  5. Return True or False.

class Student:
    def __init__(self,marks):
        self.marks = marks
    def __gt__(self,other):
        return self.marks > other.marks
s1 = Student(60)
s2 = Student(55)
print(s1>s2)
```

```python
#18. Create a Rectangle class and overload the == operator.
# Concept
# Equality Operator Overloading (__eq__)
# What to do
# Compare rectangles using their area.
# How to solve
# 1. Create class Rectangle.
# 2. Store length and width.
# 3. Calculate area.
# 4. Overload __eq__().
# 5 . Compare the areas.

class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width 
    def __eq__(self,other):
        area1 = self.length * self.width 
        area2 = self.length * self.width
        return area1 == area2

r1 = Rectangle(3,4)
r2 = Rectangle(2,6)

print(r1 == r2)
```

```python
#19. Create a Money class and overload the + and - operators.
# Concept
# Multiple Operator Overloading
# What to do
# Perform addition and subtraction of money amounts.
# How to solve
# 1. Create class Money.
# 2. Store the amount.
# 3. Overload __add__().
# 4. Overload __sub__().
# 5. Return new Money objects after each operation.

class Money:
    def __init__(self,amount):
        self.amount = amount
    def __add__(self,other):
        return Money(self.amount + other.amount)
    def __sub__(self,other):
        return Money(self.amount - other.amount)
m1 = Money(3000)
m2 = Money(400)

add = m1 + m2
print("addition :",add.amount)

sub = m1 - m2
print("subtraction :", sub.amount)
```

```python
#20. Create a Distance class and overload <, >, and ==.
# Concept
# Comparison Operator Overloading
# What to do
# Compare two distance objects.
# How to solve
# 1. Create class Distance.
# 2. Store the distance value.
# 3. Overload:
# o __lt__() for <
# o __gt__() for >
# o __eq__() for ==
# 4. Compare the stored distance values.
# 5. Return True or False.

class Distance:
    def __init__(self,distance):
        self.distance = distance
    def __lt__(self,other):
        return self.distance < other.distance
    def __gt__(self,other):
        return self.distance > other.distance 
    def __eq__(self,other):
        return self.distance == other.distance
d1 = Distance(80)
d2 = Distance(90)
print(d1 < d2)
print(d1 > d2)
print(d1 == d2)
```
