```python
#7. Write a function manual_deep_copy(lst) that manually creates a deep copy of a nested list
# without using copy.deepcopy().

def manual_deep_copy(lst):
    if isinstance(lst, list):
        return [manual_deep_copy(item) for item in lst]
    else:
        return lst


original = [1, [2, 3], [4, [5, 6]], 7]
copied = manual_deep_copy(original)

print("Original:", original)
print("Copied  :", copied)

copied[1][0] = 99
copied[2][1][0] = 100

print("\nAfter modification:")
print("Original:", original)
print("Copied  :", copied)

```

```python
#9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a
# nested element and checking the original. 


def detect_copy_type(original, copied):
    if isinstance(original[1], list):
        original[1][0] = "TEST"
    elif isinstance(original[2], dict):
        original[2]["name"] = "TEST"

    if (isinstance(copied[1], list) and copied[1][0] == "TEST") or \
       (isinstance(copied[2], dict) and copied[2].get("name") == "TEST"):
        return "Shallow Copy"
    else:
        return "Deep Copy"


import copy

data = [10, [20, 30], {"name": "Ram"}, 40]

shallow_copy = copy.copy(data)   # shallow
deep_copy = copy.deepcopy(data)  # deep

print("Shallow detection:", detect_copy_type(data[:], shallow_copy))
print("Deep detection   :", detect_copy_type(data[:], deep_copy))
```

```python
#25. Online Shopping Cart (Interview Level)
# Attributes
# • customer_name
# • cart_total
# • wallet_balance
# Methods
# 1. add_item(price)
# 2. remove_item(price)
# 3. checkout()
# Rules
# • Cart total cannot become negative.
# • Checkout only if wallet balance is enough.
# • Deduct wallet balance after successful checkout.
# • Cart becomes zero after payment.
# • Print suitable messages for every case.


class Online_Shopping_Cart:
    def __init__(self, customer_name, cart_total, wallet_balance):
        self.customer_name=customer_name
        self.cart_total=cart_total
        self.wallet_balance=wallet_balance
    def add_item(self,price):
        self.cart_total +=price
        print("Added item:",price)
    def remove_item(self,price):
        if self.cart_total - price>=0:
            self.cart_total -= price
            print("Removed item:",price)
        else:
            print("Cannot remove item")
    def checkout(self):
        if self.cart_total==0:
            print("cart is empty:")
        elif self.wallet_balance>=self.cart_total:
            self.wallet_balance -= self.cart_total
            print("checkout successful:",self.cart_total)
            print("Remaining wallet:",self.wallet_balance)
            self.cart_total=0
            print("cart is now empty:")
        else:
            print("Insufficient balance:")

cart1 =Online_Shopping_Cart("Rajesh",20,2000)
cart1.add_item(500)     
cart1.add_item(800)     
cart1.remove_item(300)  
cart1.checkout() 
```
