```python
#1.Class Name: BankAccount
# Problem
# Create a class named BankAccount.
# Explanation
# A bank account stores money. The balance should not be directly accessible from outside
# the class, so create a private attribute named __balance.
# Implement the following methods:
# • deposit(amount)
# o Deposit money into the account.
# o Amount must be greater than 0.
# • withdraw(amount)
# o Withdraw money.
# o Withdrawal should happen only if:
# ▪ amount > 0
# ▪ amount <= available balance

class Bank_account:
    def __init__(self,bal):
        self.__bal=bal
    def deposit(self,amount):
        self.__bal+=amount
        print("Money depsited:",amount)
        print("total balance:",self.__bal)
    def withdraw(self,amount):
        if amount>0 and amount<=self.__bal:
            print("withdraw money:",amount)
            print("remaining balance:",self.__bal)

bank = Bank_account(2000)
bank.deposit(200)
bank.withdraw(200)
```

```python
#10.Class Name: Account
# Problem
# Create class Account.
# Explanation
# Private attribute
# __balance
# Whenever balance changes,
# store every balance value inside a class variable
# Example
# history = []
# Methods
# • deposit()
# • withdraw()
# Every successful update should append the latest balance to history.

class Account:
    history = []
    def __init__(self,balance):
        self.__balance=balance
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            Account.history.append(self.__balance)
            print(f"Deposited{amount}, new balance : {self.__balance}")
        else:
            print("Deposit amount must be positive")
    def withdraw(self,amount):
        if amount>0 and amount <= self.__balance:
            self.__balance = amount
            Account.history.append(self.__balance)
            print(f"withdraw{amount},new balance :{self.__balance}")
        else:
            print("Invalid withdraw amount:")
            
    def get_deposit(self):
        return self.__balance
    @classmethod
    def show_history(cls):
        return cls.history
acc=Account(22000)
acc.deposit(200)
acc.withdraw(2000)
acc.get_deposit()
acc.show_history()
```

```python
#16.Class Name: Shape
# Problem
# Create an abstract class Shape.
# Explanation
# Declare abstract methods
# • calculate_area()
# • calculate_perimeter()
# Create subclasses like
# • Circle
# • Rectangle
# Implement both methods.


from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def calaculate_area(self):
        pass
    @abstractmethod
    def calculate_perimeter(self):
        pass
class Circle(Shape):
    def __init__(self,r):
        self.r = r
    def calaculate_area(self):
        area = 3.14 * self.r *self.r
        print("Area of circle=",area)
    def calculate_perimeter(self):
        perimeter =2 *3.14*self.r
        print("Perimeter of circle=",perimeter)
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def calaculate_area(self):
        area = self.length * self.width
        print("Area of Reactangle=",area)
    def calculate_perimeter(self):
        perimeter= 2* (self.length + self.width)
        print("Perimeter of Rectangle =",perimeter)
c=Circle(5)
c.calaculate_area()
c.calculate_perimeter()

r=Rectangle(6,6)
r.calaculate_area()
r.calculate_perimeter()
    
```

```python
#25. Class Name: Shape3D
# Problem
# Create abstract class Shape3D.
# Explanation
# Abstract methods
# • volume()
# • surface_area()
# Implement in
# • Cube
# • Sphere
# • Cylinder
# Each uses different formulas.
import math

class Shape3D(ABC):
    @abstractmethod
    def volume(self):
        pass
    @abstractmethod
    def surface_area(self):
        pass
class Cube(Shape3D):
    def __init__(self,side):
        self.side = side
    def volume(self):
        return self.side**3
    def surface_area(self):
        return 6*(self.side**2)
class Sphere(Shape3D):
    def __init__(self, radius):
        self.radius = radius
    def volume(self):
        return(4/3)* math.pi *(self.radius**3)
    def surface_area(self):
        return 4 * math.pi *(self.radius **2)
class Cylinder(Shape3D):
    def __init__(self,radius,height):
        self.radius = radius 
        self.height = height
    def volume(self):
        return math.pi * (self.radius**2)*self.height
    def surface_area(self):
        return math.pi *self.radius*(self.radius + self.height)
c = Cube(3)
print("Cube Volume:", c.volume())
print("Cube Surface Area:", c.surface_area())

s = Sphere(4)
print("Sphere Volume:", s.volume())
print("Sphere Surface Area:", s.surface_area())

cy = Cylinder(6,8)
print("Cylinder Volume:", cy.volume())
print("Cylinder Surface Area:", cy.surface_area())
```

```python
#30. Class Name: FileHandler
# Problem
# Create abstract class FileHandler.
# Explanation
# Abstract methods
# • read()
# • write(data)
# Implement in
# • TextFileHandler
# • CSVFileHandler
# Each subclass should define how it reads from and writes to its respective file type.
from abc import ABC, abstractmethod
import csv

class FileHandler(ABC):
    @abstractmethod
    def read(self):
        pass
    @abstractmethod
    def write(self,data):
        pass
class TextFileHandler(FileHandler):
    def __init__(self,filename):
        self.filename = filename
        
    def read(self):
        with open(self.filename,"r")as file:
            content = file.read()
        print("Read the Textfile:")
        print(content)
        return content
        
    def write(self,data):
        with open(self.filename,"w")as file:
            file.write(data)
        print("write the Textfile:")
        
class CSVFileHandler(FileHandler):
    def __init__(self,filename):
        self.filename = filename
        
    def read(self):
        with open(self.filename, "r") as file:
            reader = csv.reader(file)
        print("Read the CSVfile:")
        for row in reader:
            print(row)
                
    def write(self,data):
        with open(self.filename,"w",newline =" ") as file:
            writer = csv.writer(file)
            writer.writerows(data)
        print("writes the CSVfile:")

t = TextFileHandler("notes.txt")
t.read()
t.write("Hello")

c = CSVFileHandler("data.csv")
c.read()
c.write(["name"])
```
