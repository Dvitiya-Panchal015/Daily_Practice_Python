```python
#23. Take multiple lines of text from user until they type “STOP”, then save all lines in a file.


n = int(input("How many lines do you want to enter:"))

lines = []

for i in range(n):
    line = input()
    lines.append(line)

with open("student.txt","w") as file:
    for i in lines:
        file.write(i+"\n")

print("All lines saved sucessfully to student.txt")
```

```python
#24. Write a program to encrypt the content of a file (simple character shifting) and save it in
# another file.


with open("student.txt","r") as file:
    content = file.read()
encrypt = " "
for ch in content:
    encrypt += chr((ord(ch)+3)%256)
with open("encrypt.txt","w") as file:
    file.write(encrypt)
print("File encrypted sucessfully.")
```

```python
#25. Read a file containing student marks, calculate average, and write the result in another
# file.

total = 0
count = 0

try:
    with open("marks.txt", "r") as file:
        for line in file:
            if line.strip():
                total += int(line.strip())
                count += 1

    average = total / count if count > 0 else 0

    with open("average.txt", "w") as file:
        file.write("Average Marks: " + str(average))

    print("Average calculated and saved sucessfully.")
except FileNotFoundError:
    print("Error: marks.txt file is not found.")
    
```

```python
#26. Create a program that keeps a record of daily expenses in a file and can show total
# expenses.

with open("student.txt","a")as file:
    expense = float(input("Enter today expense:"))
    file.write(str(expense)+"\n")
print("Expensse saved")
total = 0 
with open("student.txt","r") as file:
    for line in file:
        line = line.strip()
        if line:
            try:
                total += float(line)
            except ValueError:
                print(f"skipping invalid entry:{line}")
print("total expense:",total)
    
```

```python
#27. Write a program to remove all blank lines from a file.


with open("student.txt","r") as file:
    lines = file.readlines()
with open("student.txt","w") as file:
    for line in lines:
        if line.strip():
            file.write(line)
print("Blank lines removed sucessfully.")
```

```python
#28. Read a file and write only the unique lines into another file (remove duplicates).

with open("student.txt","r")as file:
    lines = file.readlines()
unique_lines = set(lines)

with open("unique.txt","w")as file:
    for line in unique_lines:
        file.write(line)
print("unique lines saved sucessfully.")
```

```python
#29.Create a menu-driven program:
# • Write to file
# • Read file
# • Append to file
# • Exit


while True:
    print("1. write to file")
    print("2. Read file")
    print("3.Append to file")
    print("4. Exit")
    choice = int(input("Enter your choice:"))
    if choice == 1:
        with open("student.txt","w")as file:
            data = input("Enter text to write:")
            file.write(data+"\n")
        print("Data written successfully.")
    elif choice == 2:
        try:
            with open("student.txt","r")as file:
                content = file.read()
                print("\nFile conten:\n",content)
        except FileNotFoundError:
            print("file not found.")
    elif choice == 3:
        with open("student.txt","a")as file:
            data = input("Enter text to append:")
            file.write(data + "\n")
        print("data append successfully.")
    elif choice == 4:
        print("Exiting program...")
        break
    else:
        print("Invalid choice")
```
