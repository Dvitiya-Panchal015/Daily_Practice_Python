```python
#1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
# from both B and C).
# Concept
# Hybrid Inheritance (Combination of Single + Multiple Inheritance)
# What to do
# • Create parent class A.
# • Create class B inheriting from A.
# • Create another class C.
# • Create class D inheriting from both B and C.
# How to solve
# 1. Create class A.
# 2. Add one method.
# 3. Create class B(A).
# 4. Create class C.
# 5. Create class D(B, C).
# 6. Create an object of D.
# 7. Call methods from all parent classes.



class A:
    def displayA(self):
        print("This is class A")

class B(A):
    def displayB(self):
        print("This is class B")

class C:
    def displayC(self):
        print("This is class C")

class D(B, C):
    def displayD(self):
        print("This is class D")

o = D()

o.displayA()
o.displayB()
o.displayC()
o.displayD()
```

```python
#2.Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
# Concept
# Real-world Hybrid Inheritance
# What to do
# • Person is the base class.
# • Employee inherits Person.
# •  Student inherits Person.
# • TeachingAssistant inherits Employee and Student.
# How to solve

# Hierarchy
# Person
# / \
# Employee Student
# \ /
# TeachingAssistant
# Steps
# 1. Store name in Person.
# 2. Employee stores salary.
# 3. Student stores marks.
# 4. TeachingAssistant stores department.
# 5. Display all information.

class Person:
    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Name: {self.name}")


class Employee(Person):
    def __init__(self, salary,**kwargs):
        super().__init__(**kwargs)
        self.salary = salary

    def display(self):
        super().display()
        print(f"Salary: {self.salary}")


class Student(Person):
    def __init__(self, marks,**kwargs):
        super().__init__(**kwargs)
        self.marks = marks

    def display(self):
        super().display()
        print(f"Marks: {self.marks}")


class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        super().__init__(name = name , salary = salary, marks = marks )
        self.department = department

    def display(self):
        Employee.display(self)
        Student.display(self)
        print(f"Department: {self.department}")


ta1 = TeachingAssistant("Rohit", 25000, 88, "Computer Science")

print("---- Teaching Assistant ----")
ta1.display()
   
```

```python
#3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
# Concept
# Hybrid Inheritance
# What to do
# • Vehicle is parent.
#  • Car inherits Vehicle.
# • Electric is another parent.
# • ElectricCar inherits Car and Electric.
# How to solve
# Hierarchy
# Vehicle
# |
# Car Electric
# \ /
# ElectricCar

# Steps
# 1. Vehicle has start().
# 2. Car has drive().
# 3. Electric has battery().
# 4. ElectricCar accesses all methods.

class Vehicle:
    def start(self):
        print("vehicle started")
class Car(Vehicle):
    def drive(self):
        print("Car is driving:")
class Electric:
    def battery(self):
        print("Battery is charging")
class ElectricCar(Car,Electric):
    def __init__(self,brand):
        self.brand = brand
    def show(self):
        print(f"Electric Car : {self.brand}")


e = ElectricCar("Tesla 3")
e.show()
e.drive()
e.battery()
v = Vehicle()
v.start()
```

```python
#4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
# Concept
# Method Resolution Order (MRO)
# What to do
# Create
# A
# |
# B

# C

# D(B,C)
# Print
# D.mro()
# How to solve
# 1. Create four classes.
# 2. Give same method name in multiple classes.
# 3. Print ClassName.mro().
# 4. Observe Python's searching order.

class A:
    def show(self):
        print("Class A")
class B(A):
    def show(self):
        print("Class B")
class C:
    def show(self):
        print("Class c")
class D(B,C):
    def show(self):
        print("Class D")

        
d = D() 
d.show()

print(D.mro())
```

```python
#5. Create class A, B, C, D and solve the Diamond Problem using super().
# Concept
# Diamond Problem
# Hierarchy
# A
# / \
# B C
# \ /
# D
# What to do
# Use super() everywhere.
# How to solve
# 1. Create constructor in A.
# 2. Call super() in B.
# 3. Call super() in C.
# 4. Call super() in D.
# 5. Create object of D.
# 6. Observe A constructor runs only once.

class A:
    def __init__(self):
        print("A")
class B(A):
    def __init__(self):
        super().__init__()
        print("B")
class C(A):
    def __init__(self):
        super().__init__()
        print("C")
class D(B,C):
    def __init__(self):
        super().__init__()
        print("D")
d = D()
print(D.mro())
```

```python
#6. Create a banking system using hybrid inheritance.
# Concept
# Hybrid + Real-world Example
# What to do
# Account
# |
# Savings

# Loan

# SavingsLoanAccount

# How to solve
# 1. Account stores account number.
# 2. Savings stores balance.
# 3. Loan stores loan amount.
# 4. SavingsLoanAccount inherits both.
# 5. Print all details.


# Base class
class Account:
    def __init__(self, acc_no, **kwargs):
        super().__init__(**kwargs)
        self.acc_no = acc_no
        print("Account constructor called")

class Savings(Account):
    def __init__(self, acc_no, balance, **kwargs):
        super().__init__(acc_no=acc_no, **kwargs)
        self.balance = balance
        print("Savings constructor called")

class Loan(Account):
    def __init__(self, acc_no, loan_amount, **kwargs):
        super().__init__(acc_no=acc_no, **kwargs)
        self.loan_amount = loan_amount
        print("Loan constructor called")

class SavingsLoanAccount(Savings, Loan):
    def __init__(self, acc_no, balance, loan_amount):
        super().__init__(acc_no=acc_no, balance=balance, loan_amount=loan_amount)
        print("SavingsLoanAccount constructor called")

    def display(self):
        print(f"Account Number: {self.acc_no}")
        print(f"Balance: ₹{self.balance}")
        print(f"Loan Amount: ₹{self.loan_amount}")


customer = SavingsLoanAccount("ACC12345", 50000, 200000)

print("\n---- Savings + Loan Account ----")
customer.display()
print("\nMRO:", SavingsLoanAccount.mro())
```

```python
#7.Create Shape, Color, Circle, and ColoredCircle.
# Concept
# Hybrid Inheritance
# What to do
# Shape
# |
# Circle

# Color

# ColoredCircle
# How to solve
# 1. Shape has area().
# 2. Color has color().
# 3. Circle inherits Shape.
# 4. ColoredCircle inherits Circle and Color.
# 5. Display area and color.

class Shape:
    def area(self):
        print("calculate area of shapes:")
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * (self.radius**2)
class Color:
    def __init__(self,color):
        self.color = color
    def get_color(self):
        return self.color
class ColoredCircle(Circle,Color):
    def __init__(self,radius,color):
        Circle.__init__(self,radius)
        Color.__init__(self,color)
    def display(self):
        print(f"circle area :{self.radius}")
        print(f"circle color : {self.color}")

c = ColoredCircle(6,"Yellow")
c.display()
```

```python
#8. Create constructors at every level.
# Concept
# Constructor Chaining

# What to do
# Every class has its own constructor.
# How to solve
# 1. Create constructor in every class.
# 2. Use super() everywhere.
# 3. Print constructor execution order.
# 4. Observe execution using MRO.

class Shape:
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        print("Shape constructor called")


class Circle(Shape):
    def __init__(self, radius, **kwargs):
        super().__init__(**kwargs)
        self.radius = radius
        print("Circle constructor called")


class Color:
    def __init__(self, color, **kwargs):
        super().__init__(**kwargs)
        self.color = color
        print("Color constructor called")


class ColoredCircle(Circle, Color):
    def __init__(self, radius, color, **kwargs):
        super().__init__(radius=radius, color=color, **kwargs)
        print("ColoredCircle constructor called")

    def display(self):
        print(f"Circle radius: {self.radius}")
        print(f"Circle color: {self.color}")


cc = ColoredCircle(5, "Blue")

print("\nColored Circle ")
cc.display()

print("\nMRO:", ColoredCircle.mro())
```

```python
#9. Override methods in intermediate classes.
# Concept
# Method Overriding
# What to do
# A
# |
# B

# C

# D(B,C)
# All classes have
# show()
# How to solve
# 1. Create same method.
# 2. Override in B.
# 3. Override in C.
# 4. Call from D.
# 5. Observe MRO.

class A:
    def show(self):
        print("class A")
class B(A):
    def show(self):
        print("class B")
class C:
    def show(self):
        print("class C")
class D(B,C):
    def show(self):
        print("class D")
d = D()
d.show()

print(D.mro())
```

```python
#11. Duck Typing Questions (11-20)
# What is Duck Typing? Explain with a simple example.
# Concept
# Duck Typing
# "If an object behaves like a duck, Python treats it like a duck."
# What to do
# Create two unrelated classes having the same method.
# How to solve
# 1. Create Dog.
# 2. Create Robot.
# 3. Both have speak().

# 4. Write one function calling obj.speak().
# 5. Pass both objects.

class Dog:
    def speak(self):
        print("Dog says bhauuu bhauuu...")
class Robot:
    def speak(self):
        print("Robot says beep beeppp...")
def make_it_speak(obj):
    obj.speak()

d = Dog()
d.speak()

r = Robot()
r.speak()
make_it_speak(d)
make_it_speak(r)
```

```python
#15. Create Circle and Rectangle both having area().
# Concept
# Duck Typing
# What to do
# Function
# calculate(obj)
# How to solve
# 1. Circle implements area().
# 2. Rectangle implements area().
# 3. Function calls area().
# 4. Print returned value.

import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)


class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


def calculate(obj):
    print("Area:", obj.area())


c = Circle(5)
r = Rectangle(10, 4)
calculate(c)   
calculate(r)  
```

```python
#18.Create Logger classes using log(message).
# Concept
# Duck Typing
# What to do
# Two unrelated logger classes.
# How to solve
# 1. FileLogger → log().
# 2. ConsoleLogger → log().
# 3. Function accepts logger object.
#4. Calls log(message). 

class FileLogger:
    def log(self, message):
        print(f"[FileLogger] Writing to file: {message}")


class ConsoleLogger:
    def log(self, message):
        print(f"[ConsoleLogger] Printing to console: {message}")


def process_log(logger, message):
    logger.log(message)   


file_logger = FileLogger()
console_logger = ConsoleLogger()

process_log(file_logger, "System started successfully.")
process_log(console_logger, "User logged in.")
```

```python
#19.Create a hybrid inheritance structure and apply Duck Typing.
# Concept
# Hybrid + Duck Typing

# What to do
# Create hybrid inheritance.
# Create another unrelated class with same method.
# How to solve
# 1. Create hybrid hierarchy.
# 2. Final child has method show().
# 3. Another unrelated class also has show().
# 4. One function calls show() for both.

class A:
    def show(self):
        print("Class A")
class B(A):
    def show(self):
        print("Class B")
class C:
    def show(self):
        print("Class C")
class D(B,C):
    def show(self):
        print("Class D")
class Unrelated:
    def show(self):
        print("Unrelated class show()")
        
def display(obj):
    obj.show()

d = D()
u = Unrelated()

display(d)
display(u)

print("\n mro:",D.mro())
```

```python
#20. Create Employee hierarchy using Hybrid Inheritance and use Duck Typing.
# Concept
# Hybrid Inheritance + Duck Typing
# What to do
# Hierarchy
# Person
# |
# Employee

# Manager

# HR

# TeamLead
# Another class
# Freelancer
# Both TeamLead and Freelancer have
# work()
# How to solve

# 1. Build hybrid inheritance.
# 2. Final child defines work().
# 3. Freelancer also defines work().
#4. Write a function:
# d def do_work(obj):
# obj.work()
# 5. Pass both objects to the function.
# 6. Observe that the same function works for unrelated classes because of Duck Typing.


class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name):
        super().__init__(name)

class Manager(Employee):
    pass

class HR(Employee):
    pass

class TeamLead(Manager, HR):
    def work(self):
        print(self.name)
class Freelancer:
    def __init__(self, name):
        self.name = name

    def work(self):
        print(self.name)

def do_work(obj):
    obj.work()

t = TeamLead("Rohan")
f = Freelancer("Priya")

do_work(t)
do_work(f)  
```
