```python
#23.Drop the original Duration column after creating the numerical duration
# features.

df = df.drop(columns = ["Duration"])
df
```

```python
#24.Replace non-stop in Total_Stops with 0 stops.

df.replace({'Total_Stops':{'non_stop':0}},inplace = True)
df
```

```python
#25.Extract only the numerical value from Total_Stops and convert the column into
# an integer data type. Handle missing or invalid values appropriately.

df["Total_Stops"] = pd.to_numeric(df["Total_Stops"], errors = "coerce")
df["Total_Stops"]
```

```python
df["Total_Stops"] = df["Total_Stops"].fillna(0).astype(int)
df["Total_Stops"]
```

```python
#26.Split the Route column using → as the separator and create Route_1,
# Route_2, Route_3, Route_4, and Route_5.

route_split = df['Route'].str.split('→', n=4, expand=True)

route_split.columns = ['Route_1', 'Route_2', 'Route_3', 'Route_4', 'Route_5']

df[['Route_1', 'Route_2', 'Route_3', 'Route_4', 'Route_5']] = route_split

print(df[['Route_1', 'Route_2', 'Route_3', 'Route_4', 'Route_5']])
```

```python
#27.Identify the missing values created in Route_3, Route_4, and Route_5 after
# splitting the route. 


missing_values = df[["Route_3","Route_4","Route_5"]].isnull().sum()
missing_values
```

```python
#28.Replace the missing values in the route columns with an appropriate value
# such as None.

df[['Route_3', 'Route_4', 'Route_5']] = (
    df[['Route_3', 'Route_4', 'Route_5']].fillna('None'))

df[['Route_3', 'Route_4', 'Route_5']]
```

```python
#29.Drop the original Route column after creating and cleaning the route features.

df.drop("Route", axis = 1, inplace = True)
df.head()
```
