```python
#30. Define class Rectangle with private __length and __width. Add getter/setter and
# area method.

class Rectangle:
    def __init__(self,length,width):
        self.__length=length
        self.__width=width
    def get_display(self):
        print(self.__length)
        print(self.__width)
    def set_display(self,new_length,new_width):
        if new_length>0:
            self.__length = new_length
            print("length updated:",self.__length)
            if new_width>0:
                self.__width=new_width
                print("width updated:",self.__width)
            else:
                print("width is not updated:")
        else:
            print("length is not updated:")
    def area(self):
        return self.__length*self.__width
        
ret=Rectangle(20,30)
ret.set_display(20,30)
ret.area()
        
```
