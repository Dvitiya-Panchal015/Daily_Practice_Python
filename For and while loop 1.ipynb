```python
#18. Print the Fibonacci series up to 10 terms using while loop.


n= 10 
a,b=0,1
count=0
while count<n:
    print(a, end=" ")
    next_num = a+b
    a=b
    b=next_num
    count+=1
```

```python
#19.#19. Use while loop to check if a given number is prime or not.

num = 1
while num<=10:
    if num>1:
        for i in range(2,num):
            if num%i==0:
                print("not prime:")
                break
            else:
                print("Prime Number:")
    else:
        print("Not a prime no.:")
    num+=1
```

```python
#22. (Hard) Take a number n from the user. Using while loop, print all prime numbers between
#2 and n.

# Prime numbers between 2 and n using while loop

n = int(input("Enter a number: "))

num = 2   
while num <= n:
    is_prime = True  
    i = 2
    while i < num:   
        if num % i == 0:
            is_prime = False   
            break
        i += 1

    if is_prime:
        print(num, "is Prime")
    else:
        print(num, "is Not Prime")

    num += 1




    
```

```python
#24. (Hard) Take two numbers start and end from the user. Using while loop, print all
#Armstrong numbers between start and end. (Example: 153 is Armstrong because 1³ + 5³ + 3³
#= 153)

start = int(input("Enter a num:"))
end = int(input("Enter a num:"))

num=start 
while num<=end:
    digits=str(num)
    power = len(digits)
    total=0
    i=0
    while i<power:
        total+=int(digits[i])**power
        i+=1
    if total==num:
        print(num,"is Armstrong")
    num+=1
```

```python
#25. (Hard) Create a number guessing game using while loop:
#• Generate a random number between 1 to 100.
#• Give the user 7 chances to guess.
#• After each guess, tell whether the number is "Too High" or "Too Low".
#• If user guesses correctly, print "Congratulations!" and stop.
#• If chances are over, reveal the number.

import random


secret_number = random.randint(1, 100)

chances = 7   
attempt = 1

while attempt <= chances:
    guess = int(input(f"Attempt {attempt}: Enter your guess (1-100): "))

    if guess == secret_number:
        print("Congratulations! You guessed it right!")
        break
    elif guess > secret_number:
        print("Too High!")
    else:
        print("Too Low!")

    attempt += 1



if attempt > chances and guess != secret_number:
    print("Sorry, chances over! The number was:", secret_number)
```
