```python
#24.(Hard) Take a student's marks in 5 subjects. Calculate percentage and print grade
#according to these rules:
#• 90 or above → A+
#• 80–89 → A
#• 70–79 → B
#• 60–69 → C
#• Below 60 → Fail Also print "Passed with Distinction" if percentage >= 75.

for i in range(5):
    marks = int(input("Enter the student marks:"))

if(marks>=90):
    print("A")
elif(marks>=80):
    print("B")
elif(marks>=70):
    print("C")
else:
    print("Fail")
```
