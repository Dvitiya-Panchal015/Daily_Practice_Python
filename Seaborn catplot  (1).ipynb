```python
# A. Titanic Dataset — 15 Questions
# Use:
# import seaborn as sns
# titanic = sns.load_dataset("titanic")
# 1. Survival by Gender
# Use catplot() to show the number of passengers who survived and did not survive for each
# gender.
# • X-axis → sex
# • Hue → survived
# • Chart → count

import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")
titanic
```

```python
sns.catplot(data = titanic,
            x ="sex" ,
            hue = "survived",
            kind = "count")
```

```python
# 2. Survival by Passenger Class
# Create a categorical plot showing the number of passengers in each class.
# Use:
# • x="class"
# • kind="count"

sns.catplot(data = titanic,
            x ="class" ,
            kind = "count")
```

```python
# 3. Survival Count by Class
# Show the number of survivors and non-survivors for each passenger class.
# Use:
# x="class"
# hue="survived"
# kind="count"


sns.catplot(data = titanic,
            x = "class",
            hue = "survived",
            kind = "count")
```

```python
# 4. Survival Percentage by Gender
# Create a categorical visualization comparing survival across sex.
# Use catplot() and choose an appropriate categorical plot.
# Then identify which gender has a higher survival rate.

sns.catplot(data = titanic,
            x = "sex",
            y ="survived",
            kind = "bar")
```

```python
# 5. Age Distribution by Gender
# Use catplot() to compare passenger age across different genders.
# Try an appropriate kind for showing the distribution.

sns.catplot(data = titanic,
            x = "age",
            y ="sex",
            kind = "box")
```

```python
# 6. Age Distribution by Passenger Class
# Create a categorical plot showing age distribution for:
# • First class
# • Second class
# • Third class


sns.catplot(data = titanic,
            x = "class",
            y ="age",
            palette = "dark",
            kind = "box")
```

```python
# 7. Fare by Passenger Class
# Use catplot() to compare fare across the three passenger classes.
# Add:
# • Appropriate color
# • Title
# • X/Y labels

sns.catplot(data = titanic,
            x = "class",
            y ="age",
            palette = "pastel",
            kind = "box")
plt.title("Passenger class")
plt.xlabel("Class")
plt.ylabel("Age")
```

```python
# 8. Fare Distribution by Gender
# Create a categorical visualization of fare for male and female passengers.
# Identify whether the fare distributions appear different.

sns.catplot(data = titanic,
            x = "sex",
            y = "fare",
            palette = "pastel",
            kind = "box")
plt.title("Fare Distribution")
plt.xlabel("sex")
plt.ylabel("Fare")
```

```python
# 9. Age vs Passenger Class
# Use a categorical plot to compare the distribution of age across class.
# Add sex using hue.

sns.catplot(data = titanic,
            x = "class",
            y = "age",
            hue = "sex",
            palette = "husl",
            kind = "box")
plt.title("Age v/s Passenger")
plt.xlabel("age")
plt.ylabel("passenger")
```

```python
# 10. Survival by Class and Gender
# Create a categorical visualization showing survival for each passenger class and gender.
# Use:
# x="class"
# hue="sex"
# Use an appropriate kind.

sns.catplot(data = titanic,
            x = "class",
            y = "survived",
            hue = "sex",
            palette = "husl",
            kind = "box")
plt.title("Survived by class and gender")
plt.xlabel("class")
plt.ylabel("survived")
```

```python
# 11. Embark Town vs Survival
# Create a count plot using catplot() showing passenger counts for each embark_town.
# Use survived as hue.

sns.catplot(data = titanic,
            x = "embark_town",
            hue = "survived",
            palette = "husl",
            kind = "box")
plt.title("Embark Town vs Survived")
plt.xlabel("embark_town")
plt.ylabel("class")
```

```python
# 12. Passenger Class vs Fare with Gender
# Create a categorical visualization comparing fare across passenger classes.
# Use:
# x="class"
# y="fare"
# hue="sex"
# Choose an appropriate kind.

sns.catplot(data = titanic,
            x = "class",
            y = "fare",
            hue = "sex",
            palette = "husl",
            kind = "box")
plt.title("Passenger class vs fare with gender")
plt.xlabel("class")
plt.ylabel("fare")
```

```python
#13. Survival vs Age Group
# Create an age_group column:
# • Child → Age < 18
# • Adult → 18–59
# • Senior → 60+
# Use catplot() to compare survival across these groups.

import seaborn as sns
import matplotlib.pyplot as plt

titanic["age_group"] = titanic["age"].apply(
    lambda age: "Child" if age < 18 else ("Adult" if age < 60 else "Senior")
)

sns.catplot(
    data=titanic,
    x="age_group",
    y="survived",
    kind="bar"
)

plt.title("Survival by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# 14. Class, Gender and Survival
# Create a categorical visualization where:
# • X → class
# • Y → survival-related measure
# • Hue → sex
# • Columns → survived
# Use col= to create separate panels.

sns.catplot(data = titanic,
            x = "class",
            y = "fare",
            hue = "sex",
            palette = "husl",
            col = "survived",
            kind = "box")
plt.title("class, gender and survived")
plt.xlabel("class")
plt.ylabel("fare")
```

```python
# 15. Titanic Multi-Panel Analysis
# Create a catplot() with:
# • X → class
# • Y → fare
# • Hue → sex
# • Columns → survived
# Use an appropriate categorical kind.
# Add:
# • Figure size
# • Title
# • Color palette
# • Appropriate labels
# Explain what you observe.

sns.catplot(data = titanic,
            x = "class",
            y = "fare",
            hue = "sex",
            palette = "husl",
            col = "survived",
            kind = "box")
plt.title("titanic multi- panel analysis")
plt.xlabel("class")
plt.ylabel("fare")
```

```python
# 30. Final Seaborn catplot() Challenge
# Using the Tips dataset, create a professional categorical visualization with:
# x="day"
# y="total_bill"
# hue="sex"
# col="time"
# Choose the most appropriate kind.
# Customize:
# • palette
# • height
# • aspect
# • col_wrap if appropriate
#• Title
# •  Axis labels
#• Legend
# T hen answer:
#1. Which day has the highest typical bill?
# 2 . Is there a difference between male and female customers?
# 3. Does the pattern change between lunch and dinner?
# 4. Which combination of day and time appears to have the highest bills?


sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    col = "time",
    palette="rocket",
    height = 5,
    aspect = 0.5,
    kind="boxen"

)
plt.title("Multi-Panel Tips Analysis")
plt.xlabel("day")
plt.ylabel("total_bill")
plt.legend("bill")
plt.show()
```
