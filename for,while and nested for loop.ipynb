```python
#10. Keep taking numbers from user until they enter 0, then print sum.
# Explanation: Use while True, break when 0 entered.
# Input: 5 7 3 0
# Output: Sum = 15


total = 0   

while True:   
    num = int(input("Enter a number (0 to stop): "))
    
    if num == 0:   
        break      
    
    total += num   

print("Sum =", total)
```

```python
#11. Reverse a number using while loop.
# Explanation: Use modulo and division to reverse digits.
# Input: 1234
# Output: 4321

num = 1234
rev=0
while num>0:
    digit = num%10
    rev = rev*10+digit
    num = num//10
print(rev)
```

```python
#12.Check if a number is prime using while loop.
# Explanation: Check divisibility from 2 to sqrt(n).
# Input: 17
# Output: Prime Number


num = int(input("Enter a number:"))

if num<2:
    print("Not prime")
else:
    i=2
    is_prime = True
    while i*i <= num:
        if num % i == 0:
            is_prime = False
            break
        i+=1
    if is_prime:
        print("Prime number:")
    else:
        print("Not prime:")
```

```python
#13. Print Fibonacci series up to n terms using while loop.
# Explanation: Use two variables a=0, b=1 and generate next terms.
# Input: 8
# Output: 0 1 1 2 3 5 8 13


n= int(input("Enter a no.:"))
a,b=0,1
count=0

while count<n:
    print(a, end=" ")
    next_num = a+b
    a=b
    b= next_num
    count+=1

```

```python
# 16. Find sum of digits of a number using while loop.
# Explanation: Use modulo 10 and integer division.
# Input: 12345
# Output: Sum of digits = 15

num = int(input("Enter a number:"))

total = 0
while num>0:
    digit = num%10
    total += digit
    num = num//10
print("sum of digits=", total)
```

```python
#19. Print prime numbers between 1 to 100 using nested for loop.
# Explanation: Outer loop for numbers, inner loop to check divisibility.
# Input: None
# Output: List of primes


for num in range (2,100):
    is_prime=True

    for i in range(2, num):
        if num%i==0:
            is_prime=False
            break
    if is_prime:
        print(num,end=" ")
```

```python
#22. Print a hollow square pattern of size n using nested loop.
# Input: 5
# Output: Square with only border *

n= int(input("Enter size:"))
for i in range(n):
    for j in range(n):
        if i==0 or i==n-1 and j==0 or j==n-1:
            print("*", end= " ")
        else:
            print(" ", end=" ")
    print()

```
