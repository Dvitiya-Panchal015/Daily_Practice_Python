```python
#19.Take a paragraph. Split it into sentences (using .), strip each sentence, and join them with "\n- ".

#split
paragraph= "   python is powerful.programming language .that is easy to learn.    "
paragraph=paragraph.split(".")
print(paragraph)

#strip
sentences = [s.strip() for s in paragraph if s.strip()]
print(sentences)
```
