```python
#9. Create a class ComplexNumber and overload +, -, *, and == operators completely.

class ComplexNumber:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return ComplexNumber(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return ComplexNumber(self.x - other.x, self.y - other.y)
    def __mul__(self,other):
        return ComplexNumber(self.x * other.x, self.y * other.y)
    def __eq__(self,other):
        return ComplexNumber(self.x == other.x, self.y == other.y)
    def __str__(self):
        return f"{self.x} + {self.y}i"
        
    def display(self):
        print(f"{self.x} + {self.y}i")
        
p1 = ComplexNumber(3,8)
p2 = ComplexNumber(4,5)

print("Addition:", p1 + p2)
print("Subtraction:",p1 - p2)
print("Multiplication:",p1 * p2)
print("equal:", p1==p2)

```

```python
#10. Create a class Student with private marks list.
# • Add setter with validation (each mark 0-100)
# • Calculate average
# • Overload > operator to compare two students based on average

class Student:
    def __init__(self,name,marks):
        self.name = name
        self.__marks = []
    def set_marks(self,marks):
        if all(0 <= m <= 100 for m in marks):
            self.__marks = marks
        else:
            raise ValueError("Each mark must be between 0 and 100.")
    def average(self):
        if len(self.__marks)==0:
            return 0
        return sum(self.__marks) / len(self.__marks)
    def __gt__(self,other):
        return self.average() > other.average()

s1 = Student("Alice",77)
s2 = Student("Bob",55)


s1.set_marks([33,44,55,66,77,88])
s2.set_marks([34,66,77,88,99,67])

print(f"{s1.name} Average:",s1.average())
print(f"{s2.name} Average:", s2.average())
if s1 > s2:
    print(f"{s1.name} has a higher average than {s2.name}")
else:
    print(f"{s2.name} has a higher average than {s2.name}")
```

```python
#15. Create a class Time (hours, minutes, seconds) and overload + operator to add
# two time objects with proper carry-over.

class Time:
    def __init__(self,hours,minutes,seconds):
        self.hours = hours
        self.minutes = minutes 
        self.seconds = seconds
    def __add__(self,other):
        n_hours = self.hours + other.hours
        n_minutes = self.minutes + other.minutes
        n_seconds = self.seconds + other.seconds

        if n_seconds >= 60:
            n_seconds -= 60
            n_minutes += 1
        if n_minutes >= 60:
            n_minutes -= 60
            n_hours += 1
        n_hours = n_hours % 24

        return Time(n_hours,n_minutes,n_seconds)
    def __str__(self):
        return f"{self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}"

t1 = Time(2,55,50)
t2 = Time(6,49,50)

t3 = t1 + t2
print("Time 1 :",t1)
print("Time 2:", t2)
print("sum:", t3)
        
```

```python
#18. Write a program using nested loops that prints a diamond pattern of numbers
# and stops printing if user enters a negative number (use break).


num = int(input("Enter a positive number:"))

if num < 0:
    print("Negative number entered.")
else:
    for i in range(1,num+1):
        for j in range(num - i):
            print(" ", end=" ")
        for j in range(1, i+1):
            print(j,end = " ")
        print()
    for i in range(num - 1, 0, -1):
        for j in range(num - i):
            print(" ", end = "")
        for j in range(1, i+1):
            print(j,end = " ")
        print()
```
